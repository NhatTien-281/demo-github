```python
import torch
import torch.nn as nn
from sklearn.datasets import make_blobs
import numpy as np
import plotly.express as px
import matplotlib.pyplot as plt
```

```python
x_np,y_nb=make_blobs(n_samples=400,cluster_std=1.4,centers=2,random_state=2)
X = torch.from_numpy(x_np).float()
y = torch.from_numpy(y_nb).float().view(-1, 1)
```

```python
X
```

```python
y
```

```python
plt.scatter(x=X[:,0],y=X[:,1],c=y_nb,cmap='Accent')
```

```python
class LogisticRegression(nn.Module):
    def __init__(self,input_dim):
        super(LogisticRegression,self).__init__()
        self.linear=nn.Linear(input_dim,1)
    def forward(self,x):
        y_hat=torch.sigmoid(self.linear(x))
        return y_hat
model=LogisticRegression(input_dim=2)
```

```python
# 1. Chọn hàm mất mát và bộ tối ưu
criterion = nn.BCELoss() # Binary Cross Entropy chuyên cho phân loại
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)

# 2. Vòng lặp huấn luyện (Training Loop)
epochs = 5000
all_losses = []

for epoch in range(epochs):
    # Forward pass: Dự đoán kết quả
    y_pred = model(X)
    
    # Tính toán lỗi
    loss = criterion(y_pred, y)
    all_losses.append(loss.item())
    
    # Backpropagation: "Thám tử tìm lỗi" và "Sửa lỗi"
    optimizer.zero_grad() # Xóa bảng (gradient) cũ
    loss.backward()      # Tìm lỗi (tính gradient)
    optimizer.step()     # Sửa lỗi (cập nhật w, b)
    
    if (epoch + 1) % 500 == 0:
        print(f'Epoch [{epoch+1}/{epochs}], Loss: {loss.item():.4f}')
        print(f"Parameter = {model.state_dict()}")


# Vẽ biểu đồ giảm Loss
plt.plot(all_losses)
plt.title("Đồ thị giảm sai số (Loss)")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.show()
```

```python
all_losses
```

```python
# Chuyển mô hình về chế độ đánh giá (không tính gradient nữa)
with torch.no_grad():
    # Lấy kết quả dự đoán (dạng xác suất 0 -> 1)
    predictions = model(X)
    
    # "Trọng tài chốt hạ": Nếu xác suất > 0.5 thì là lớp 1, ngược lại lớp 0
    predicted_classes = (predictions > 0.5).float()

# In thử 10 kết quả đầu tiên
print("10 dự đoán đầu tiên:\n", predicted_classes[:10].view(-1))
print("10 nhãn thực tế đầu tiên:\n", y[:10].view(-1))

# Vẽ biểu đồ kết quả phân loại của máy
plt.scatter(x_np[:, 0], x_np[:, 1], c=predicted_classes.numpy().reshape(-1), cmap='coolwarm')
plt.title("Kết quả phân loại của mô hình PyTorch")
plt.show()

# In ra trọng số (w) và bias (b) mà máy đã học được
for name, param in model.named_parameters():
    print(f"{name}: {param.data}")
```
