```python
obj=['Location', 'WindGustDir', 'WindDir9am', 'WindDir3pm',
       'RainToday']
num=['MinTemp', 'MaxTemp', 'Rainfall', 'Evaporation', 'Sunshine',
       'WindGustSpeed', 'WindSpeed9am', 'WindSpeed3pm', 'Humidity9am',
       'Humidity3pm', 'Pressure9am', 'Pressure3pm', 'Cloud9am', 'Cloud3pm',
       'Temp9am', 'Temp3pm']
target='RainTomorrow'
len(obj)+len(num)
```

```python
from sklearn.impute import SimpleImputer
imputer_num=SimpleImputer(missing_values=np.nan,strategy='mean')
data = df.copy()
data[num]=imputer_num.fit_transform(data[num])
data[num].info()
```

```python
imputer_obj=SimpleImputer(missing_values=np.nan,strategy='most_frequent')
data[obj]=imputer_obj.fit_transform(data[obj])
data[obj].info()
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression().fit(data_train_by_year[input_name_col],data_train_by_year[target])
```

```python
val_predic=model.predict(data_val_by_year[input_name_col])
```

```python
check_predict=pd.Series(val_predic).map({'No':0,'Yes':1})
check_predict_val=data_val_by_year[target].map({'No':0,'Yes':1})
def MAE(x,y):
    return np.mean(abs(x-y))
MAE(check_predict,check_predict_val)
```
